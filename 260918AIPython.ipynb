```python
#100분은 몇시간 몇분인지 출력해 보세요
print(100//60,'시간',100%60,'분')
```

```python
#100분 뿐만 아니라 변수를 사용해서 x분을 계산하도록 하라
Time = int(input('시간[분] 입력하시오 : '))
print(Time,'분 은',Time//60,'시간',Time%60,'분')
print(f'{Time}분 은 {Time//60}시간 {Time%60}분 입니다')
```

```python
#두 변수의 내용을 교환하자
a=int(input('첫번째 값을 입력하시오 : '))
b=int(input('두번째 값을 입력하시오 : '))
buf=a
a=b
b=buf
print('바꿨습니다 : ',a,b)
```

```python
#temp나 buffer 없이 할수있는 파이썬만의 연산
a=int(input('첫번째 값을 입력하시오 : '))
b=int(input('두번째 값을 입력하시오 : '))
a,b = b,a
print('바꿨습니다 : ',a,b)
```
