1부터 100까지의 합을 출력하시오

```python
sum = 0
i = 0
for i in range(1,101):
  sum += i

print(sum)
```

```python
def allsum(a):
    sum = 0
    for i in range(1,a+1):
        sum += i
    return(sum)

print(allsum(100))
```

정수하나 입력받아서 그 수의 구구단을 x X 1 = Y 형식으로 출력하라

```python
def product(a):
  for i in range(1, 10):
    print(f"{a} X {i} = {a * i}")

product(int(input("정수 입력하면 구구단 보여드림 : ")))
```

2단부터 9단까지 구구단을 7 x 1 = 7 형식으로 출력하시오

```python
def product(a):
  for i in range(1, 10):
    print(f"{a} X {i} = {a * i}")

for j in range(2,10):
  print(f"==={j}단===")
  product(j)
```

while문으로 1부터 100까지 합 구하기

```python
sum = 0
i = 1
while (i <= 100):
  sum += i
  i += 1

print(sum)
```

```python
# break 쓰기
sum = 0
i = 1
while (True):
  sum += i
  if i == 100:
    break
  i += 1
print(sum)
```

양의 정수를 입력받아 합계를계속 구하다가 0을 입력하면 반복을 끝내고 합계를 출력

```python
sum = 0
while (True):
  a = int(input("양의 정수 입력 : "))
  if a < 0:
    print("양의 정수를 입력하시오")
    continue
  if a == 0:
    break
  sum += a
print(sum)
```

```python
#코드5-21
del sum
scores = [85, 95, 70, 75, 99, 93, 88]

print('최댓값:', max(scores))
print('최솟값:', min(scores))
print('합:', sum(scores))
print('평균:', sum(scores)/len(scores))
```

```python
total=0
scores = [85, 95, 70, 75, 99, 93, 88]
for i in scores:
  total += i
print('합:', total)
```

```python
#max 구현하기
max_score = 0
for i in scores:
  if i > max_score:
    max_score = i
print(max_score)
```
