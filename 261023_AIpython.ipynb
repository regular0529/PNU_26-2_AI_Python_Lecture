```python
#20개의 국가에 대한 수도 데이터를 딕셔너리로 만들어줘
```

```python
capitals = {
    '대한민국': '서울', '일본': '도쿄', '중국': '베이징', '미국': '워싱턴 D.C.',
    '영국': '런던', '프랑스': '파리', '독일': '베를린', '이탈리아': '로마',
    '스페인': '마드리드', '캐나다': '오타와', '호주': '캔버라', '브라질': '브라질리아',
    '인도': '뉴델리', '러시아': '모스크바', '이집트': '카이로', '태국': '방콕',
    '베트남': '하노이', '튀르키예': '앙카라', '아르헨티나': '부에노스아이레스', '멕시코': '멕시코시티',
}

score = 0
for country, capital in capitals.items():
    an = input(country + '의 수도는? ').strip()
    if an == capital:
        print('정답')
        score += 1
    elif an == "q":
        break
    else:
        print('오답 정답은 ',capital)

print(f'{len(capitals)}문제 중 정답{score}')
```
