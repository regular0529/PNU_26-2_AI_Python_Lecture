```python
from sklearn import datasets
iris = datasets.load_iris()
```

```python
iris.keys()
```

```python
print('데이터:\n', iris['data'][:5])          # 샘플 5개
print('특성 이름:', iris['feature_names'])
print('타깃:\n', iris['target'][:5])
print('타깃 이름:', iris['target_names'])
print('데이터 크기:', iris['data'].shape)
print('타깃 크기:', iris['target'].shape)
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(iris['data'], iris['target'])
```

```python
from sklearn.neighbors import KNeighborsClassifier

knn = KNeighborsClassifier(n_neighbors=3)    # 모델 설정
knn.fit(X_train, y_train)                    # 학습
```

```python
X_new = [[6, 3.5, 5, 1.5]]            # 꽃받침 길이 6, 너비 3.5, 꽃잎 길이 5, 너비 1.5
pred = knn.predict(X_new)
print('예측 값:', pred, '->', iris['target_names'][pred])
```

```python
y_pred = knn.predict(X_test)
print('테스트 세트 예측 값:\n', y_pred)
print('테스트 세트 실제 타깃:\n', y_test)
```
